# Constraint Satisfaction Problem : Graph coloring


## Create Data

We can create a planar graph with $n$ vertices by randomly placing $n$ points in 2-dimensional Euclidean space and then performing a [Delaunay triangulation](https://en.wikipedia.org/wiki/Delaunay_triangulation).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import Delaunay

np.random.seed(1111)

# create points and sort them by 
n = 10
points = np.random.randint(100, size=(n, 2))
o = np.argsort(points[:,0])
points = points[o,:]


# triangulate
tri = Delaunay(points)


# plot
plt.triplot(points[:,0], points[:,1], tri.simplices, color = "gray")
plt.plot(points[:,0], points[:,1], 'o', color = "green", markersize = 20)

for i in range(len(points)):
       plt.annotate(i, points[i,:], 
        color='white', fontsize="large", weight='heavy',
        horizontalalignment='center', verticalalignment='center')

plt.show()
```

The triangulation can be converted into a list with an array of neighbor vertex indices for each point (see man page for [scipy.spatial.Delaunay](https://docs.scipy.org/doc/scipy/reference/generated/scipy.spatial.Delaunay.html)). 

```python
(indptr, indices) = tri.vertex_neighbor_vertices

neighbors = []

for k in range(len(indptr)-1):
    neighbors.append(np.sort(indices[indptr[k]:indptr[k+1]]))
    
neighbors
```

The first row are the indices for the points neighboring point 0.

## Define CSP

The problem is defined as a dictionary with variable names, a dictionary with domain values for each variable and a set of binary  not-equal constraints (tuples of variables that are not allowed to have the same value).  

```python
variables = [str(var) for var in range(n)]

domain = ['red', 'blue', 'green', 'orange']
#domain = ['red', 'blue', 'green']
domains = {}
for v in variables: 
    domains[v] = domain

# create binary constraints
constraints = {}
for i in range(len(neighbors)):
    for j in neighbors[i]:
        if(i<j): constraints[tuple([str(i), str(j)])] = True
constraints = constraints.keys()

csp = {'variables': variables, 'domains': domains, 'constraints': constraints, }
csp
```

## Check Completeness and Consistency

`assignment` is a dictionary. Assigned variables have an entry with the variable name as the key. `{}` is therefore the empty assignment.

```python
def complete(assignment, csp):
    return(all([v in assignment.keys() for v in csp['variables']])) 

print("complete({}, csp) =", complete({}, csp))

print("complete({...complete list...}, csp) =", 
      complete({'0': 'red', '1': 'red', '2': 'red', '3': 'red', '4': 'red', 
          '5': 'red', '6': 'red', '7': 'red', '8': 'red', '9': 'red'}, csp))
```

```python
def consistent(assignment, csp):
    for constr in csp['constraints']:
        if(constr[0] in assignment.keys() and constr[1] in assignment.keys()):
            if assignment[constr[0]] == assignment[constr[1]]: return(False)
        
    return(True)
        
print("consistent({}, csp) =", consistent({}, csp))
print("consistent({'0': 'red', '1': 'red', '2': 'blue'}, csp) =", 
      consistent({'0': 'red', '1': 'red', '2': 'blue'}, csp))
```

```python
def select_unassigned_var(assignment, csp):
    unassigned = [v for v in csp['variables'] if v not in assignment]
    return min(unassigned, key=lambda var: len(csp['domains'][var]))

def order_domain(var, assignment, csp):
    def count_conflicts(val):
        count = 0
        for v1, v2 in csp['constraints']:
            if var == v1 and v2 not in assignment and val in csp['domains'][v2]:
                count += 1
            elif var == v2 and v1 not in assignment and val in csp['domains'][v1]:
                count += 1
        return count
    return sorted(csp['domains'][var], key=count_conflicts)

def forward_check(var, val, assignment, csp):
    for v1, v2 in csp['constraints']:
        neighbor = None
        if var == v1: neighbor = v2
        elif var == v2: neighbor = v1
        if neighbor and neighbor not in assignment:
            if val in csp['domains'][neighbor]:
                csp['domains'][neighbor].remove(val)
                if not csp['domains'][neighbor]: return False
    return True

def backtrack_search(csp):
    global COUNT
    COUNT = 0
    assignment = backtrack({}, csp)
    print(f"Checked nodes: {COUNT}")
    return assignment

def backtrack(assignment, csp):
    global COUNT
    COUNT += 1
    if complete(assignment, csp): return assignment
    var = select_unassigned_var(assignment, csp)
    original_domains = {v: list(csp['domains'][v]) for v in csp['variables']}
    for val in order_domain(var, assignment, csp):
        assignment[var] = val
        if consistent(assignment, csp) and forward_check(var, val, assignment, csp):
            result = backtrack(assignment, csp)
            if result: return result
        assignment.pop(var)
        csp['domains'] = {v: list(original_domains[v]) for v in csp['variables']}
    return None
```

```python
# Hill Climbing with Min-Conflicts

def min_conflicts(csp, max_steps=1000, max_restarts=20):
    import random
    variables = csp['variables']
    domains = csp['domains']
    constraints = csp['constraints']
    restarts = 0
    steps = 0

    def conflicts(var, val, assignment):
        count = 0
        for v1, v2 in constraints:
            if var == v1 and v2 in assignment and assignment[v2] == val:
                count += 1
            elif var == v2 and v1 in assignment and assignment[v1] == val:
                count += 1
        return count

    while restarts < max_restarts:
        assignment = {v: random.choice(domains[v]) for v in variables}
        for step in range(max_steps):
            conflicted = [v for v in variables if conflicts(v, assignment[v], assignment) > 0]
            if not conflicted: return assignment, steps, restarts
            var = random.choice(conflicted)
            assignment[var] = min(domains[var], key=lambda val: conflicts(var, val, assignment))
            steps += 1
        restarts += 1
    return None, steps, restarts
```

```python
def run_experiments(sizes=[5, 10, 15], colors=['red', 'blue', 'green'], trials=3):
    import time
    results = []
    for n in sizes:
        for trial in range(trials):
            points = np.random.randint(100, size=(n, 2))
            tri = Delaunay(points)
            (indptr, indices) = tri.vertex_neighbor_vertices
            neighbors = [np.sort(indices[indptr[k]:indptr[k+1]]) for k in range(len(indptr)-1)]
            variables = [str(i) for i in range(n)]
            domains = {v: list(colors) for v in variables}
            constraints = {tuple(sorted([str(i), str(j)])): True for i in range(len(neighbors)) for j in neighbors[i] if i < j}
            csp = {'variables': variables, 'domains': domains, 'constraints': constraints.keys()}
            avg_edges = len(constraints)

            global COUNT
            COUNT = 0
            start = time.time()
            res_bt = backtrack_search(csp)
            bt_time = time.time() - start

            start = time.time()
            res_mc, steps, restarts = min_conflicts(csp)
            mc_time = time.time() - start

            results.append({
                'n': n,
                'edges': avg_edges,
                'bt_assignments': COUNT,
                'bt_time': bt_time,
                'mc_steps': steps,
                'mc_restarts': restarts,
                'mc_time': mc_time
            })
    return results
```

```python
# Tạo dữ liệu thử nghiệm
results = run_experiments(sizes=[5, 10, 15], colors=['red', 'blue', 'green', 'orange'], trials=3)

# Vẽ biểu đồ hiệu suất
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(results)
grouped = df.groupby('n').mean()

fig, axs = plt.subplots(2, 2, figsize=(12, 8))

axs[0,0].plot(grouped.index, grouped['edges'], marker='o')
axs[0,0].set_title("Số ràng buộc trung bình")
axs[0,0].set_xlabel("Số đỉnh (n)")
axs[0,0].set_ylabel("Số cạnh")

axs[0,1].plot(grouped.index, grouped['bt_assignments'], marker='o', label='Backtracking')
axs[0,1].set_title("Số lần gán biến (Backtracking)")
axs[0,1].set_xlabel("n")

axs[1,0].plot(grouped.index, grouped['mc_steps'], marker='o', label='Min-Conflicts Steps')
axs[1,0].plot(grouped.index, grouped['mc_restarts'], marker='x', label='Restarts')
axs[1,0].set_title("Hiệu suất Min-Conflicts")
axs[1,0].set_xlabel("n")
axs[1,0].legend()

axs[1,1].plot(grouped.index, grouped['bt_time'], marker='o', label='Backtracking')
axs[1,1].plot(grouped.index, grouped['mc_time'], marker='x', label='Min-Conflicts')
axs[1,1].set_title("Thời gian chạy")
axs[1,1].set_xlabel("n")
axs[1,1].legend()

plt.tight_layout()
plt.show()
```
